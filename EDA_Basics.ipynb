**Q1. What is the average, median, and standard deviation of house prices?**

```python
df['Price'].mean()
df['Price'].median()
df['Price'].std()

print('Average Price:',df['Price'].mean())
print('Median Price:',df['Price'].median())
print('Standard Deviation Price:',df['Price'].std())
```

**Q2. Which number of bedrooms is most common?**

```python
df['number of bedrooms'].mode()
```

**Q3. Check if the price data is skewed.**

```python
df['Price'].skew()
print("Price data is positively right skewed:", df['Price'].skew())
```

```python
df['Price'].plot(kind='hist')
plt.title('Distribution of Price')
plt.xlabel('Frequency')
plt.ylabel('Price')
plt.show()
```

**Q4. What is the average price for each number of bedrooms?**

```python
avg_price_per_bedroom = df.groupby('number of bedrooms')['Price'].mean()
print(avg_price_per_bedroom)
```

**Q5.  What is the relationship between living area and price? (Use : Area without Basement).**

```python
correlation = df['Area of the house(excluding basement)'].corr(df['Price'])
correlation
```

```python
plt.scatter(df['Area of the house(excluding basement)'], df['Price'])
plt.xlabel('Area without Basement')
plt.ylabel('Price')
plt.title('Relationship between Area without Basement and Price')
plt.show()
```

**Q6. Identify any anomalies where houses have high prices but low areas.**

```python
#Identify any anomalies where houses have high prices but low areas.
High_price = df['Price'].quantile(0.99)
Low_area = df['Area of the house(excluding basement)'].quantile(0.01)
```

```python
anomalies = df[(df['Price'] > High_price) & (df['Area of the house(excluding basement)'] < Low_area)]
```

**Q7. Compare average price based on number of floors and houses with or without waterfront.**

```python
avg_price = df.groupby(['number of floors','waterfront present'])['Price'].mean()
avg_price
```

**Q8. Identify the minimum and maximum house price. What does this indicate?**

```python
max_price = df['Price'].max()
min_price = df['Price'].min()

print('Maximum Price:',max_price)
print('Minimum Price:',min_price)
```

The large difference between minimum and maximum prices indicates that the dataset contains houses with very different price ranges, possibly beacuase of area, loaction, bedrooms,waterfronts,conditions,etc.

**Q9. Which location (zipcode/area) has the highest average price?**

```python
df.groupby('Postal Code')['Price'].mean().idxmax()
```
